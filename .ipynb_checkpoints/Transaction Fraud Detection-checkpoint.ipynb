In the **PaySim dataset**, the main fields **used for fraud detection** are those that describe the **transaction details and account balances**, because fraud is typically detected by **anomalous behavior in these fields**. Let’s go through them one by one:

---

## **1. `type` (Transaction Type)**

* Values: `CASH_OUT`, `TRANSFER`, `PAYMENT`, `DEBIT`, `CASH_IN`
* Fraud is more likely in **CASH_OUT** or **TRANSFER** transactions, especially if the amount is unusually large.

---

## **2. `amount` (Transaction Amount)**

* The actual money transferred in the transaction.
* Extremely high or unusual amounts can indicate fraud.

---

## **3. `oldbalanceOrg` (Original Balance of Sender)**

* Balance of the origin account **before** the transaction.
* Fraudsters may try to transfer more than the available balance or manipulate it in unusual ways.

---

## **4. `newbalanceOrig` (New Balance of Sender)**

* Balance of the origin account **after** the transaction.
* If `newbalanceOrig` is inconsistent with the transaction amount, it could indicate fraud.

---

## **5. `oldbalanceDest` (Original Balance of Receiver)**

* Balance of the destination account **before** the transaction.
* Helps detect if the destination account is unusual (e.g., new account or zero balance).

---

## **6. `newbalanceDest` (New Balance of Receiver)**

* Balance of the destination account **after** the transaction.
* Inconsistencies may indicate stolen funds are being transferred to a “mule” account.

---

## **7. `isFraud` (Target Variable)**

* `0` = Not Fraud
* `1` = Fraud
* This is **what the model tries to predict**.

---

### **Optional / Derived Features**

Sometimes you can create additional features to improve detection:

* `amount_ratio = amount / oldbalanceOrg` → fraction of balance transferred
* `balance_diff_origin = oldbalanceOrg - newbalanceOrig`
* `balance_diff_dest = newbalanceDest - oldbalanceDest`
* Time-based features (like transaction step per user, frequency)

---

✅ **Summary:**
The key predictive fields for fraud detection are:

`type`, `amount`, `oldbalanceOrg`, `newbalanceOrig`, `oldbalanceDest`, `newbalanceDest`

All others (like `step`) can be used optionally for temporal or sequential analysis.

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=100)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(f"--- {LogisticRegression} ---")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, model.predict_proba(X_test)[:,1]))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print("\n")
```
